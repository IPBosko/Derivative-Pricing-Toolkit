```python
import numpy as np
import scipy as sp
```

```python
def call_payoff(S_T, K):

    return np.maximum( S_T - K , 0.0 )

def put_payoff(S_T, K):

    return np.maximum( K - S_T , 0.0 )

def binary_call_payoff(S_T, K, Q):
    
    return np.where(S_T > K, Q, 0.0)

def binary_put_payoff(S_T, K, Q):
    
    return np.where(S_T < K, Q, 0.0)
```

```python
S_array = np.array([80, 100, 125])
K = 100
Q = 10
```

```python
print('Call payoff        ', call_payoff(S_array, K) )
print('Put payoff         ', put_payoff(S_array, K) ) 
print('Binary call payoff ', binary_call_payoff(S_array, K, Q) )
print('Binary put payoff  ', binary_put_payoff(S_array, K, Q) )
```

```python
def arithm_average(array):

    return array.mean()

def geom_average(array):

    return np.exp( arithm_average( np.log (array) ) )

def asian_call_payoff(prices_array, K, kind="arithmetic"):

    arithm = np.maximum( arithm_average(prices_array) - K , 0.0 )
    geom = np.maximum( geom_average(prices_array) - K , 0.0 )
    
    return arithm if kind == "arithmetic" else geom
```

```python
prices_array = np.array([104.0, 97.0, 92.0, 89.0, 95.0, 103.0])
K = 95.0
```

```python
print('Arithmetic Average of Stock Prices:', f"{arithm_average(prices_array):.2f}")
print('Geometric Average of Stock Prices: ', f"{geom_average(prices_array):.2f}")

print('Arithmetic Asian Call Payoff:', f"{asian_call_payoff(prices_array, K, kind="arithmetic"):.2f}")
print('Geometric Asian Call Payoff: ', f"{asian_call_payoff(prices_array, K, kind="geometric"):.2f}")
print('European Call Payoff: ', call_payoff(prices_array, K))
```
